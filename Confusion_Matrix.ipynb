```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset=pd.read_csv('place.csv')
dataset.head()
```

```python
x=dataset.iloc[:, :-1]
y=dataset['placed']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)*100
```

```python
from sklearn.metrics import confusion_matrix,precision_score,recall_score,f1_score      
```

```python
confusion_matrix(y_test, model.predict(x_test)) 
```

```python
sns.heatmap(confusion_matrix(y_test, model.predict(x_test)), annot=True, fmt='d', cmap='Blues')
```

```python
precision_score(y_test, model.predict(x_test))*100
```

```python
recall_score(y_test, model.predict(x_test))*100
```

```python
f1_score(y_test, model.predict(x_test))*100
```
