```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset=pd.read_csv('Social_Network_Ads.csv')
```

```python
dataset.head(1)
```

```python
dataset.isnull().sum()
```

```python
dataset.shape
```

```python
x=dataset.iloc[:, :-1]
y=dataset["Purchased"]
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x=scaler.fit_transform(x)
```

```python
x=pd.DataFrame(x, columns=dataset.columns[:-1])
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=0)  
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(max_depth=3)#pruning here by max depth
```

```python
dt.fit(x_train, y_train)
```

```python
dt.score(x_test, y_test) *100  
```

```python
plot_decision_regions(x.to_numpy(),y.to_numpy(),clf=dt)
```

```python
dt.predict([[35,20000]])
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(50, 50))
plot_tree(dt)
plt.savefig("tree.jpg")
plt.show()
```

```python
dt1=DecisionTreeClassifier(criterion="entropy")
dt1.fit(x_train,y_train)
```

```python
dt1.score(x_test,y_test)*100
```

```python
dt1.score(x_train,y_train)*100
```

```python
for i in range(1,20):
    dt2=DecisionTreeClassifier(max_depth=i)
    dt2.fit(x_train,y_train)
    print(dt2.score(x_train,y_train),dt2.score(x_test,y_test),i)
```
