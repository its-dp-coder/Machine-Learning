```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
```

```python
dataset=pd.read_csv('place.csv')
dataset.head()
```

```python
# sns.kdeplot(data=dataset, x='cgpa', hue='placed', fill=True)
```

```python
sns.scatterplot(data=dataset, x='cgpa', y='placement_exam_marks', hue='placed')
```

```python
x=dataset.iloc[:, :-1]
y=dataset["placed"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB
gnb=GaussianNB()
gnb.fit(x_train,y_train)    


```

```python
gnb.score(x_test,y_test)*100
```

```python
mnb=MultinomialNB()
mnb.fit(x_train,y_train)
```

```python
mnb.score(x_test,y_test)*100
```

```python
br=BernoulliNB()
br.fit(x_train,y_train)
```

```python
br.score(x_test,y_test)*100
```

```python
plot_decision_regions(x.to_numpy(), y.to_numpy(), clf=gnb)
```
